```python
import pandas as pd
import numpy as np
```

```python
results = pd.read_csv("results.csv")
results = results.fillna('110')
results
```

```python
results.groupby("ordering").quantile([0.1, 0.5, 0.9])*100
```

```python
cols_to_exclude = ["C->T", "T->C", 'global_masking_rate', 'ordering', 'mapping', 'entropy']
vals = (
    results[results["ordering"] == "110_ct"]
    .drop(columns=cols_to_exclude)
    .to_numpy()
    .ravel()
)

np.quantile(vals, [0.1, 0.5, 0.9])*100
```

```python
cols_to_exclude = ["C->G", "G->C", 'global_masking_rate', 'ordering', 'mapping', 'entropy']
vals = (
    results[results["ordering"] == "110_gc"]
    .drop(columns=cols_to_exclude)
    .to_numpy()
    .ravel()
)

np.quantile(vals, [0.1, 0.5, 0.9])*100
```

```python
cols_to_exclude = ["A->G", "G->A", 'global_masking_rate', 'ordering', 'mapping', 'entropy']
vals = (
    results[results["ordering"] == "110_ga"]
    .drop(columns=cols_to_exclude)
    .to_numpy()
    .ravel()
)

np.quantile(vals, [0.1, 0.5, 0.9])*100
```

```python
cols_to_exclude = ["C->T", "G->T", 'global_masking_rate', 'ordering', 'mapping', 'entropy']
vals = (
    results[results["ordering"] == "110_cov"]
    .drop(columns=cols_to_exclude)
    .to_numpy()
    .ravel()
)

np.quantile(vals, [0.1, 0.5, 0.9])*100
```
